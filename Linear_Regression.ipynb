### Single Variable Linear Regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression

# Load the dataset
df = pd.read_csv('/content/multiple_linear_regression_dataset.csv')
print("Dataset Columns:", df.columns.tolist())
display(df.head())

# Let's assume the last column is the target (y) and we choose one feature for single variable regression
x_col = df.columns[0]
y_col = df.columns[-1]

X_single = df[[x_col]]
y = df[y_col]

# Fit single linear regression
model_single = LinearRegression()
model_single.fit(X_single, y)

# Predict
y_pred_single = model_single.predict(X_single)

# Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df[x_col], y=y, color='blue', label='Actual Data')
plt.plot(df[x_col], y_pred_single, color='red', linewidth=2, label='Regression Line')
plt.title(f'Single Linear Regression: {y_col} vs {x_col}')
plt.xlabel(x_col)
plt.ylabel(y_col)
plt.legend()
plt.grid(True)
plt.show()
```

### Multiple Linear Regression

```python
# Features (all columns except the last target column)
X_multi = df.iloc[:, :-1]

# Fit multiple linear regression
model_multi = LinearRegression()
model_multi.fit(X_multi, y)

# Predict
y_pred_multi = model_multi.predict(X_multi)

# Print coefficients
for col, coef in zip(X_multi.columns, model_multi.coef_):
    print(f"Coefficient for {col}: {coef:.4f}")
print(f"Intercept: {model_multi.intercept_:.4f}")

# Scatter plot of Actual vs Predicted values to evaluate the multiple regression model
plt.figure(figsize=(8, 5))
sns.scatterplot(x=y, y=y_pred_multi, color='purple', alpha=0.7)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linestyle='--', linewidth=2, label='Perfect Fit')
plt.title('Multiple Linear Regression: Actual vs Predicted')
plt.xlabel('Actual Values')
plt.ylabel('Predicted Values')
plt.legend()
plt.grid(True)
plt.show()
```
