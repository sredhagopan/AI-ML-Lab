### Graph Setup and Heuristics

```python
import heapq
from collections import deque

# Graph defined as adjacency list: {node: [(neighbor, cost), ...]}
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('D', 5), ('E', 1)],
    'C': [('A', 4), ('F', 3)],
    'D': [('B', 5)],
    'E': [('B', 1), ('F', 1)],
    'F': [('C', 3), ('E', 1)]
}

# Heuristic values (straight-line distance to goal 'F')
heuristic = {
    'A': 6,
    'B': 5,
    'C': 3,
    'D': 7,
    'E': 1,
    'F': 0
}
```

### Breadth-First Search (BFS)


```python
def bfs(graph, start, goal):
    queue = deque([(start, [start])]) # Queue holds (node, path)
    visited = set()                  # Track visited nodes

    while queue:
        node, path = queue.popleft() # Get next node in BFS queue
        if node == goal:
            return path             # Return path if goal is reached

        if node not in visited:
            visited.add(node)        # Mark node as visited
            for neighbor, _ in graph.get(node, []):
                if neighbor not in visited:
                    queue.append((neighbor, path + [neighbor])) # Queue neighbor
    return None

print("BFS Path:", bfs(graph, 'A', 'F'))
```

### Depth-First Search (DFS)

```python
def dfs(graph, start, goal):
    stack = [(start, [start])] # Stack holds (node, path)
    visited = set()            # Track visited nodes

    while stack:
        node, path = stack.pop() # Get most recent node
        if node == goal:
            return path          # Return path if goal is reached

        if node not in visited:
            visited.add(node)    # Mark node as visited
            for neighbor, _ in graph.get(node, []):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor])) # Stack neighbor
    return None

print("DFS Path:", dfs(graph, 'A', 'F'))
```

### Uniform Cost Search (UCS)

```python
def ucs(graph, start, goal):
    pq = [(0, start, [start])] # Priority queue of (g_score, node, path)
    visited = set()            # Track visited nodes

    while pq:
        cost, node, path = heapq.heappop(pq) # Get lowest cost node
        if node == goal:
            return path, cost               # Return path and cost

        if node not in visited:
            visited.add(node)                # Mark node as visited
            for neighbor, edge_cost in graph.get(node, []):
                if neighbor not in visited:
                    heapq.heappush(pq, (cost + edge_cost, neighbor, path + [neighbor])) # Update cost
    return None, float('inf')

path, cost = ucs(graph, 'A', 'F')
print(f"UCS Path: {path} with Cost: {cost}")
```

### Greedy Best-First Search

```python
def greedy_best_first(graph, start, goal, h):
    pq = [(h[start], start, [start])] # Priority queue of (h_score, node, path)
    visited = set()                   # Track visited nodes

    while pq:
        _, node, path = heapq.heappop(pq) # Get node with smallest heuristic
        if node == goal:
            return path                  # Return path if goal is reached

        if node not in visited:
            visited.add(node)             # Mark node as visited
            for neighbor, _ in graph.get(node, []):
                if neighbor not in visited:
                    heapq.heappush(pq, (h[neighbor], neighbor, path + [neighbor])) # Priority by h
    return None

print("Greedy Best-First Path:", greedy_best_first(graph, 'A', 'F', heuristic))
```

### A* Search

```python
def a_star(graph, start, goal, h):
    pq = [(h[start], 0, start, [start])] # Priority queue of (f_score, g_score, node, path)
    visited = set()                      # Track visited nodes

    while pq:
        _, g, node, path = heapq.heappop(pq) # Get lowest f_score node
        if node == goal:
            return path, g                   # Return path and actual cost

        if node not in visited:
            visited.add(node)                # Mark node as visited
            for neighbor, edge_cost in graph.get(node, []):
                if neighbor not in visited:
                    g_score = g + edge_cost  # Path cost
                    f_score = g_score + h[neighbor] # Estimated total cost
                    heapq.heappush(pq, (f_score, g_score, neighbor, path + [neighbor])) # Priority by f
    return None, float('inf')

path, cost = a_star(graph, 'A', 'F', heuristic)
print(f"A* Path: {path} with Cost: {cost}")
```
